```python
import random

import torch, os
from TTS.api import TTS

# Get device
device = "cuda" if torch.cuda.is_available() else "cpu"

model_name = "GPT_XTTS_v2.0_Full_7_6_SNF"

model_path = f"/cluster/data/deri/TTS/TTS_CH/trained/{model_name}/"
config_path = f"/cluster/data/deri/TTS/TTS_CH/trained/{model_name}/config.json"
# Init TTS
tts = TTS(
    model_path=model_path,
    config_path=config_path,
    progress_bar=True
).to(device)
```

```python
import sounddevice as sd
from scipy.io.wavfile import write

# Parameters
fs = 44100  # Sample rate: 44.1 kHz
seconds = 15  # Duration of recording in seconds
```

```python
print("Recording started...")
recording = sd.rec(int(seconds * fs), samplerate=fs, channels=1)  # Use channels=1 for mono, change to 2 for stereo
sd.wait()  # Wait until recording is finished
print("Recording finished.")

# Save the recording as a WAV file
write("output.wav", fs, recording)
print("Audio saved as output.wav")
```

```python
conditioning_paths = ["output.wav"]

line1 = 'Guten Morgen Bern, heute zeigt sich der Himmel wechselhaft: Am Vormittag dominieren sanfte Wolken und gelegentliche Sonnenstrahlen, während am Nachmittag vereinzelt kurze Regenschauer aufziehen könnten.'
line2 = 'Die Temperaturen bewegen sich zwischen 8 und 14 Grad, begleitet von einer leichten Brise aus südwestlicher Richtung – ein typischer, abwechslungsreicher Frühlingstag.'
dial = "ch_vs"
tts.tts_to_file(text=line1, speaker_wav=conditioning_paths, language=dial, split_sentences=False, file_path=os.path.join("from_rec_path", f'test_audio-{dial}1.wav'))
tts.tts_to_file(text=line2, speaker_wav=conditioning_paths, language=dial, split_sentences=False, file_path=os.path.join("from_rec_path", f'test_audio-{dial}2.wav'))
```
