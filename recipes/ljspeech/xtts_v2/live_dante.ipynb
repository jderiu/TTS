```python
import random

from pydub import AudioSegment
import torch, os
from TTS.api import TTS

# Get device
device = "cuda" if torch.cuda.is_available() else "cpu"

model_name = "GPT_XTTS_v2.0_LJSpeech_FT-December-02-2024_12+05PM-afa6889d"

model_path = f"/cluster/data/deri/TTS/TTS_dante/trained/{model_name}/"
config_path = f"/cluster/data/deri/TTS/TTS_dante/trained/{model_name}/config.json"
# Init TTS
tts = TTS(
    model_path=model_path,
    config_path=config_path,
    progress_bar=True
).to(device)
tts.eval()

tts.synthesizer.output_sample_rate = 24000

tts.synthesizer.seg = tts.synthesizer._get_segmenter('it')
# Run TTS
# ❗ Since this model is multi-lingual voice cloning model, we must set the target speaker_wav and language
# Text to speech list of amplitude values as output
#wav = tts.tts(text="Hello world!", speaker_wav="my/cloning/audio.wav", language="en")
# Text to speech to a file

speaker_wavs_base = 'D:/02Datasets/02 Audio Processing/dante/inferno/audio_segment_clean/canto_5'
speaker_wavs = [os.path.join(speaker_wavs_base, x) for x in os.listdir(speaker_wavs_base) if 'terzina' in x]
path = 'D:/02Datasets/02 Audio Processing/dante/purgatorio'
```

```python
speaker_wav_manual = [
     'D:/02Datasets/02 Audio Processing/dante/inferno/audio_segment_clean/canto_2\\terzina_23.wav',
     'D:/02Datasets/02 Audio Processing/dante/inferno/audio_segment_clean/canto_2\\terzina_47.wav',
     'D:/02Datasets/02 Audio Processing/dante/inferno/audio_segment_clean/canto_2\\terzina_44.wav',
     'D:/02Datasets/02 Audio Processing/dante/inferno/audio_segment_clean/canto_2\\terzina_45.wav',
]
speaker_wavs_random = random.sample(speaker_wavs, 4)

speaker_wavs_used = speaker_wav_manual

for idx, line in enumerate(lines):
    line = lines[idx]
    alternatives = 3
    for alt in range(alternatives):
        random.shuffle(speaker_wavs_used)
        tts.tts_to_file(text=line, speaker_wav=speaker_wavs_used, language="it", split_sentences=False, file_path=os.path.join(path, f'pred_audio/canto_2/terzine/audio_{idx}_{alt}.wav'))
```
